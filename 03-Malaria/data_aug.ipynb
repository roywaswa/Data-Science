```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

# 1. Load datasets
df2015 = pd.read_csv('data/children_2015.csv')
df2020 = pd.read_csv('data/children_2020.csv')

# 2. Define features and target
features = ['h22', 'ml0', 'v024', 'v025', 'v190'] #, 'hc1', 'hc27',
target   = 'ml13e'

# 3. Prepare training data (fill missing, one-hot encode)
X_train = (df2015[features]
           .fillna(0)
           .pipe(lambda d: pd.get_dummies(d, columns=['ml0','v024','v025'], drop_first=True)))
y_train = df2015[target].fillna(0)

# 4. Fit logistic regression
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

# 5. Prepare 2020 data the same way
X2020 = (df2020[features]
         .fillna(0)
         .pipe(lambda d: pd.get_dummies(d, columns=['hc27','ml0','v024','v025'], drop_first=True))
         .reindex(columns=X_train.columns, fill_value=0))

# 6. Predict and append to 2020 file
df2020['ml13e_imputed'] = model.predict(X2020)

# 7. Save augmented file
output_path = 'exports/children_2020_with_act_use.csv'
df2020.to_csv(output_path, index=False)
print(f"Augmented file written to: {output_path}")
```
