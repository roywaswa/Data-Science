## Data Extraction

```python
# Import necessary libraries
import pandas as pd
import os

# Define function to extract specific columns
def extract_columns(input_file, output_file, columns_to_keep, columns_rename={}):
    """
    Extract specific columns from a CSV file and save to a new CSV file.
    
    Parameters:
    input_file (str): Path to the input CSV file
    output_file (str): Path to save the output CSV file
    columns_to_keep (list): List of column names to keep
    columns_rename (dict): Dictionary mapping original column names to new names
    
    Returns:
    DataFrame: The extracted data
    """
    try:
        # Read the CSV file
        df = pd.read_csv(input_file)
        
        # Display the first few rows of the original dataframe
        print("First 5 rows of the original dataframe:")
        display(df.head())
        
        # Print all column names for reference
        print("\nAll columns in the CSV file:")
        print(df.columns.tolist())
        
        # Check if all specified columns exist in the dataframe
        missing_columns = [col for col in columns_to_keep if col not in df.columns]
        if missing_columns:
            print(f"\nWarning: The following columns were not found in the CSV: {missing_columns}")
            # Filter out missing columns
            columns_to_keep = [col for col in columns_to_keep if col in df.columns]
            
        if not columns_to_keep:
            print("Error: None of the specified columns were found in the CSV file.")
            return None
            
        # Select only the specified columns
        df_selected = df[columns_to_keep]
        
        # Rename columns if specified
        if columns_rename:
            # Filter out any rename mappings for columns that don't exist
            valid_renames = {old: new for old, new in columns_rename.items() if old in df_selected.columns}
            if valid_renames:
                df_selected = df_selected.rename(columns=valid_renames)
                print("\nRenamed the following columns:")
                for old, new in valid_renames.items():
                    print(f"  {old} -> {new}")
            
            # Check for invalid rename mappings
            invalid_renames = {old: new for old, new in columns_rename.items() if old not in df_selected.columns}
            if invalid_renames:
                print("\nWarning: Could not rename the following columns (not found in selection):")
                for old, new in invalid_renames.items():
                    print(f"  {old} -> {new}")
        
        # Display the first few rows of the new dataframe
        print("\nFirst 5 rows of the extracted dataframe:")
        display(df_selected.head())
        
        # Save the selected columns to a new CSV file
        df_selected.to_csv(output_file, index=False)
        
        print(f"\nSuccessfully extracted {len(columns_to_keep)} columns to {output_file}")
        return df_selected
        
    except Exception as e:
        print(f"Error: {str(e)}")
        return None
```

## Data Aggregation

```python
import pandas as pd
import numpy as np
from typing import Dict, List, Union, Callable

def aggregate_by_cluster(
    df: pd.DataFrame,
    cluster_column: str,
    category_columns: List[str] = None,
    aggregation_methods: Dict[str, Union[str, List[str], Callable]] = None
) -> pd.DataFrame:
    """
    Aggregate a dataframe by cluster with separate count columns for categorical variables.
    
    Parameters:
    -----------
    df : pandas DataFrame
        The input dataframe to be aggregated
    cluster_column : str
        The column name containing cluster assignments
    category_columns : list, optional
        List of categorical columns to create separate count columns for each unique value
    aggregation_methods : dict, optional
        Dictionary mapping column names to aggregation methods
        If a column is not specified, default aggregations will be applied
    
    Returns:
    --------
    pandas DataFrame
        Aggregated dataframe with one row per cluster and separate count columns for categories
    """
    if aggregation_methods is None:
        aggregation_methods = {}
    
    if category_columns is None:
        category_columns = []
    
    # Make a copy of the dataframe to avoid modifying the original
    data = df.copy()
    
    # Create result dataframe starting with cluster counts
    result = pd.DataFrame(data[cluster_column].value_counts().sort_index())
    result.columns = ['total_count']
    
    # Process each categorical column to create separate count columns
    for col in category_columns:
        if col in data.columns and col != cluster_column:
            # Get unique values in this column
            unique_values = data[col].dropna().unique()
            
            # For each unique value, create a count column
            for val in unique_values:
                # Create a new column with count of this value in each cluster
                count_series = data[data[col] == val].groupby(cluster_column).size()
                result[f"{col}_{val}"] = count_series
    
    # Handle the remaining columns with the standard aggregation function
    # Remove category columns from aggregation_methods
    remaining_agg_methods = {k: v for k, v in aggregation_methods.items() 
                            if k not in category_columns and k != cluster_column}
    
    # Identify remaining columns
    remaining_cols = [col for col in data.columns 
                     if col not in category_columns and col != cluster_column]
    
    if remaining_cols:
        # Create a subset dataframe with only the remaining columns
        subset_df = data[[cluster_column] + remaining_cols].copy()
        
        # Apply standard aggregation to remaining columns
        standard_aggs = {}
        
        # Identify numeric and categorical columns
        numeric_cols = subset_df.select_dtypes(include=['number']).columns.tolist()
        cat_cols = subset_df.select_dtypes(include=['object', 'category']).columns.tolist()
        
        if cluster_column in numeric_cols:
            numeric_cols.remove(cluster_column)
        if cluster_column in cat_cols:
            cat_cols.remove(cluster_column)
        
        # Apply user-specified aggregation methods
        for col, agg_method in remaining_agg_methods.items():
            if col in subset_df.columns:
                standard_aggs[col] = agg_method
        
        # Apply default aggregations for numeric columns not specified
        for col in numeric_cols:
            if col not in standard_aggs:
                standard_aggs[col] = ['mean', 'median', 'min', 'max']
        
        # Apply default aggregations for remaining categorical columns
        for col in cat_cols:
            if col not in standard_aggs:
                standard_aggs[col] = [
                    ('mode', lambda x: x.mode()[0] if not x.mode().empty else np.nan),
                    ('unique_count', 'nunique')
                ]
        
        if standard_aggs:
            # Perform standard aggregation
            standard_result = subset_df.groupby(cluster_column).agg(standard_aggs)
            
            # Flatten the column names if multi-level
            if isinstance(standard_result.columns, pd.MultiIndex):
                standard_result.columns = [f"{col[0]}_{col[1]}" if isinstance(col, tuple) else col 
                                         for col in standard_result.columns]
            
            # Join with the result dataframe
            result = result.join(standard_result)
    
    # Fill NaN values with 0 for count columns
    count_cols = [col for col in result.columns if col != 'total_count' and not col.startswith(('age_', 'stratum_', 'region_'))]
    result[count_cols] = result[count_cols].fillna(0).astype(int)
    
    return result
```

```python
def aggregate_dataset(
    df, 
    cluster_column='cluster',
    category_columns=None,
    numeric_columns=None,
    simple_agg_columns=None,
    custom_agg_methods=None,
    file_format='csv',
    **kwargs
):
    """
    Reads a dataset file and performs cluster-based aggregation with customizable settings.
    
    Parameters:
    -----------
    file_path : str
        Path to the data file
    cluster_column : str, default='cluster'
        The column name containing cluster assignments
    category_columns : list or None, default=None
        List of categorical columns to create separate count columns for each unique value
    numeric_columns : list or None, default=None
        List of numeric columns to apply numeric aggregations to
    simple_agg_columns : dict or None, default=None
        Dictionary mapping column names to simple aggregation method (e.g., {'region': 'first'})
    custom_agg_methods : dict or None, default=None
        Dictionary mapping column names to custom aggregation methods
    file_format : str, default='csv'
        Format of input file ('csv', 'excel', etc.)
    **kwargs : 
        Additional arguments to pass to the file reading function (e.g., sep=',', encoding='utf-8')
    
    Returns:
    --------
    pandas DataFrame
        Aggregated dataframe with one row per cluster
    """
    import pandas as pd
    import numpy as np
    
    # Set defaults for optional parameters
    if category_columns is None:
        category_columns = []
    if numeric_columns is None:
        numeric_columns = []
    if simple_agg_columns is None:
        simple_agg_columns = {}
    if custom_agg_methods is None:
        custom_agg_methods = {}
    
    # Convert numeric columns if specified
    for col in numeric_columns:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors='coerce')
    
    # Prepare aggregation methods
    agg_methods = {}
    
    # Add simple aggregations (like 'first', 'last', etc.)
    for col, method in simple_agg_columns.items():
        if col in df.columns and col != cluster_column:
            agg_methods[col] = method
    
    # Add custom aggregation methods
    for col, method in custom_agg_methods.items():
        if col in df.columns and col != cluster_column:
            agg_methods[col] = method
    
    # Perform the aggregation
    result = aggregate_by_cluster(df, cluster_column, category_columns, agg_methods)
    return result
```

```python
def isolate_columns(columns, year_suffix="_15"):
    """
    Isolates columns ending with the specified suffix and constant columns without numerical suffixes,
    excluding temperature columns from constants
    
    Args:
        columns (list): List of column names
        year_suffix (str): Suffix to filter for (default: "_15")
        
    Returns:
        tuple: (columns_with_suffix, constant_columns, combined_columns)
    """
    import re
    
    # Find columns ending with the specified suffix
    # Handle both formats like "_15" and just "15"
    columns_with_suffix = [col for col in columns if col.endswith(year_suffix) or 
                          (year_suffix.startswith('_') and col.endswith(year_suffix[1:]))]
    
    # Find constant columns without numerical suffix
    # Pattern to match columns ending with numbers
    pattern = re.compile(r'\d+$')
    
    # Columns that don't end with numbers
    constant_columns = [col for col in columns if not pattern.search(col)]
    
    # Remove temperature columns from the constant columns
    constant_columns = [col for col in constant_columns 
                        if not col.startswith('Temperature_')]
    
    # Combined list of both types
    combined_columns = columns_with_suffix + constant_columns
    
    return columns_with_suffix, constant_columns, combined_columns
```

```python
import pandas as pd

def rename_duplicate_columns(df):
    # Define mapping for affected columns - all names limited to 10 characters
    rename_map = {
        # Total count columns
        'total_coun': 'tot_cnt1',
        'total_co_1': 'tot_cnt2',
        'total_co_2': 'tot_cnt3',
        
        # Residence columns
        'residence_': 'res_type',
        'residenc_1': 'res_cat',
        'residenc_2': 'res_regn',
        
        # Cluster columns
        'cluster': 'cluster',  # keep as is
        'cluster_sh': 'clstr_sh',
        'cluster_hh': 'clstr_hh',
        
        # Net owned columns
        'net_owned_': 'net_own1',
        'net_owne_1': 'net_own2',
        
        # Livestock columns
        'Livestock_': 'lvstk',
        'Livestoc_1': 'lvstk_t1',
        'Livestoc_2': 'lvstk_t2',
        'Livestoc_3': 'lvstk_t3',
        'Livestoc_4': 'lvstk_t4',
        
        # Temperature columns
        'DTemp20': 'DTemp20',  # keep as is
        'Matemp20': 'MaxTmp20',
        'Meantem20': 'MeanTm20',
        'Mintem20': 'MinTmp20'
    }
    
    # Create a new DataFrame with renamed columns
    # Only rename columns that exist in the dataframe
    columns_to_rename = {col: rename_map[col] for col in df.columns if col in rename_map}
    
    if columns_to_rename:
        df = df.rename(columns=columns_to_rename)
        print(f"Renamed {len(columns_to_rename)} columns")
    else:
        print("No columns were renamed")
    
    return df

# Rename the columns
df_renamed = rename_duplicate_columns(gdf)
```

```python
mal_2015_selected = mal_2015_selected.copy()
# Proportion One: Proportion of children under 5 years of age who slept under an ITN the night before the survey
# Formula: Total number of Children who slept under an ITN (fever_rec_only treated nets) / Total number of children (alive_yes) in the cluster
mal_2015_selected['prop_children_5'] = mal_2015_selected['fever_rec_only treated nets'] / mal_2015_selected['alive_yes']

# Proportion Two: Proportion of households with at least one ITN
# Formula: Total number of households with at least one ITN (net_use_yes) / Total number of households (total_count_hh) in the cluster
mal_2015_selected['prop_households_ITN'] = mal_2015_selected['net_use_yes'] / mal_2015_selected['total_count_hh']

# Proportion Three: Proportion of individuals with access to an ITN in the household
# Formula: Total number of individuals with access to an ITN (nets_sum) / Total number of individuals (hh_size_sum) in the cluster
mal_2015_selected['prop_individuals_ITN'] = mal_2015_selected['nets_sum'] / mal_2015_selected['hh_size_sum']

# Proportion Four: Proportion of existing ITNs that were used the night before the survey
# Formula: Total number of individuals who slept under an ITN (net_use_yes) / Total number of existing ITNs (nets_sum) in the cluster
mal_2015_selected['prop_ITN_use'] = mal_2015_selected['net_use_yes'] / mal_2015_selected['nets_sum']

# Proportion Five: Proportion of individuals who slept under an ITN the night before the survey
# Formula: Total number of individuals who slept under an ITN (net_use_yes) / Total number of individuals (hh_size_sum) in the cluster
mal_2015_selected['prop_individuals_slept_ITN'] = mal_2015_selected['net_use_yes'] / mal_2015_selected['hh_size_sum']

# Proportion Six: Proportion of households with at least one ITN for every 2 people
# Formula: Total number of households with at least one ITN (net_use_yes) / (Total number of households (total_count_hh) * 2) in the cluster
mal_2015_selected['prop_households_ITN_2'] = mal_2015_selected['net_use_yes'] / (mal_2015_selected['total_count_hh'] * 2)
```
